```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
X = np.array([[1],[2],[3],[4],[5]])
y = np.array([35,40,50,60,70])
```

```python
print(X.shape)
print(y.shape)
```

```python
w=0.0
b=0.0
```

```python
def predict(X,w,b):
  return X.flatten()*w+b
```

```python
y_pred = predict(X,w,b)
print(y_pred)
```

```python
def mse(y,y_pred):
  return np.mean((y-y_pred)**2)
```

```python
print(mse(y,y_pred))
```

```python
lr =0.008
epochs = 6000
n =len(y)

losses = []

for i in range(epochs):
  y_perd = predict (X,w,b)

  dw =(2/n)*np.sum(X.flatten()*(y_perd-y))
  db =(2/n)*np.sum(y_perd-y)

  w = w-lr*dw
  b = b-lr*db

  loss = mse(y,y_perd)
  losses.append(loss)
```

```python
plt.plot(losses)
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.show()
```

```python
plt.scatter(X, y)
plt.plot(X ,predict(X,w,b))
plt.xlabel("Hours")
plt.ylabel("Marks")
plt.show()
```

```python
print("Weights:",w)
print("Bias:",b)
```
